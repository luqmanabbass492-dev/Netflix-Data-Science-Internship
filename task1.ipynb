```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
#Dataset loading
df = pd.read_csv('Dataset.csv')

print("Dataste is loaded!")
print("ship:", df.shape)
df.head()
```

```python
#missing values checking
print("missing values:")
print(df.isnull().sum())
print("\n" + "="*50 + "\n")

#check duplicate raws 
duplicates = df.duplicated().sum()
print("fduplicate raws :{duplicates}")
```

```python
# Missing values ko fill karein
df['director'] = df['director'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')
df['rating'] = df['rating'].fillna(df['rating'].mode()[0])
df['duration'] = df['duration'].fillna('Unknown')

# Baaki missing values check karein
print("Missing Values Handle Karne Ke Baad:")
print(df.isnull().sum())
```

```python
#Date column convert to Datetime
df['date_added'] = pd.to_datetime (df['date_added'], errors='coerce')

#creat new columns
df['year_added'] = df['date_added'].dt.year
df['month_added']= df['date_added'].dt.month

print('date columns made!')
df[['date_added', 'year_added', 'month_added']].head()
```

```python
# Duration ko number aur type mein split karein
df['duration_num'] = df['duration'].str.extract(r'(\d+)').astype(float)
df['duration_type'] = df['duration'].str.extract(r'([a-zA-Z]+)')

print("Duration column clean ho gaya!")
df[['duration', 'duration_num', 'duration_type']].head()
```

```python
# Type column ko numerical mein convert karein (Movie=0, TV Show=1)
df['type_encoded'] = df['type'].map({'Movie': 0, 'TV Show': 1})

print("Type Column Encode Ho Gaya:")
df[['type', 'type_encoded']].head()
```

```python
# Cleaned dataset save karein
df.to_csv('netflix_cleaned.csv', index=False)
print("Cleaned dataset save ho gaya: netflix_cleaned.csv")
```
