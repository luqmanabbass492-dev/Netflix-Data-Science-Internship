```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# Dataset load karein (same folder se)
df = pd.read_csv('netflix_cleaned.csv')

print("Dataset Loaded Successfully!")
print("Shape:", df.shape)
df.head()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='type', data=df, palette='Set2')
plt.title('Movies vs TV Shows on Netflix', fontsize=14, fontweight='bold')
plt.xlabel('Type')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
top_countries = df['country'].value_counts().head(10)
sns.barplot(x=top_countries.values, y=top_countries.index, palette='viridis')
plt.title('Top 10 Countries with Most Netflix Content', fontsize=14, fontweight='bold')
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
yearly = df['release_year'].value_counts().sort_index()
plt.plot(yearly.index, yearly.values, marker='o', color='crimson', linewidth=2)
plt.title('Netflix Content Release Trend by Year', fontsize=14, fontweight='bold')
plt.xlabel('Year')
plt.ylabel('Number of Titles')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
top_genres = df['listed_in'].value_counts().head(10)
sns.barplot(x=top_genres.values, y=top_genres.index, palette='magma')
plt.title('Top 10 Genres on Netflix', fontsize=14, fontweight='bold')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')
plt.show()
```

```python
plt.figure(figsize=(10, 7))
rating_counts = df['rating'].value_counts().head(8)
plt.pie(rating_counts.values, labels=rating_counts.index, autopct='%1.1f%%', 
        colors=sns.color_palette('pastel'), startangle=90)
plt.title('Netflix Content Rating Distribution', fontsize=14, fontweight='bold')
plt.show()
```

```python
print("="*60)
print("📊 KEY INSIGHTS FROM EDA")
print("="*60)
print(f"\n1. Total Content: {len(df)} titles")
print(f"2. Movies: {df[df['type']=='Movie'].shape[0]}")
print(f"3. TV Shows: {df[df['type']=='TV Show'].shape[0]}")
print(f"4. Top Country: {df['country'].value_counts().index[0]}")
print(f"5. Top Genre: {df['listed_in'].value_counts().index[0]}")
print(f"6. Top Rating: {df['rating'].value_counts().index[0]}")
print(f"7. Latest Release Year: {df['release_year'].max()}")
print(f"8. Earliest Release Year: {df['release_year'].min()}")
```
